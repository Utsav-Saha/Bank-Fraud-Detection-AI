Visualize Fraud vs Genuine

```python
import matplotlib.pyplot as plt

class_counts = df["Class"].value_counts()

plt.figure(figsize=(8, 5))
plt.bar(
    ["Genuine", "Fraud"],
    [class_counts[0], class_counts[1]]
)

plt.title("Genuine vs Fraudulent Transactions")
plt.xlabel("Transaction Type")
plt.ylabel("Number of Transactions")
plt.show()
```

Compare Fraud Amount vs Genuine Amount

```python
genuine_amount = df[df["Class"] == 0]["Amount"]
fraud_amount = df[df["Class"] == 1]["Amount"]

print("Average Genuine Amount:", genuine_amount.mean())
print("Average Fraud Amount:", fraud_amount.mean())

print("Median Genuine Amount:", genuine_amount.median())
print("Median Fraud Amount:", fraud_amount.median())
```

Fraud Over Time

```python
df["Time_Hour"] = (df["Time"] // 3600).astype(int)
```

calculate fraud transactions per hour:

```python
fraud_per_hour = (
    df[df["Class"] == 1]
    .groupby("Time_Hour")
    .size()
)
```

```python
plt.figure(figsize=(12, 5))

plt.plot(fraud_per_hour.index, fraud_per_hour.values)

plt.title("Fraud Transactions Over Time")
plt.xlabel("Time (Hour)")
plt.ylabel("Number of Fraud Transactions")

plt.show()
```

Correlation Analysis

```python
class_correlation = correlation["Class"].sort_values()

print(class_correlation)
```

Correlation Heatmap

Separate Fraud and Genuine Data

```python
genuine = df[df["Class"] == 0]
fraud = df[df["Class"] == 1]

print("Genuine Transactions:", len(genuine))
print("Fraud Transactions:", len(fraud))
```

Compare the Amount of Distribution

```python
plt.figure(figsize=(10, 5))

plt.boxplot(
    [genuine["Amount"], fraud["Amount"]],
    labels=["Genuine", "Fraud"]
)

plt.title("Transaction Amount: Genuine vs Fraud")
plt.ylabel("Amount")

plt.show()
```
